```python

# ==========================================
# Python 資料型別實作作業：聚餐費用試算器
# 測試一：原始資料（4 人、折價券 100 元）
# ==========================================

# 1. 輸入原始資料
event_name = "  WEEKEND DINNER  "
meal_amount = " NT$1,200 "
people = "4"

service_rate = 0.10
discount = 100
budget_per_person = 300

# 2. 整理聚餐名稱
# strip()：去除字串前後空白
# lower()：將英文字母轉成小寫
clean_event_name = event_name.strip().lower()

# 3. 整理餐費並轉換型別
# replace()：移除貨幣符號、逗號及空白
clean_meal_amount = meal_amount.replace("NT$", "").replace(",", "").strip()

# int()：將餐費文字轉換成整數
meal_amount_number = int(clean_meal_amount)

# int()：將文字人數轉換成整數
people_number = int(people)

# 4. 計算服務費
# 服務費必須以原始餐費計算
service_fee = meal_amount_number * service_rate

# 5. 計算應付總額
# 原始餐費 + 服務費 - 折價券
total_amount = meal_amount_number + service_fee - discount

# 6. 計算每人分攤費用
per_person = total_amount / people_number

# 7. 判斷是否超過每人預算
# 如果每人費用大於預算，結果為 True
over_budget = per_person > budget_per_person

# 8. 輸出費用摘要
print("========== 聚餐費用試算器 ==========")
print(f"聚餐名稱：{clean_event_name}")
print(f"原始餐費：{meal_amount_number:.2f} 元")
print(f"服務費：{service_fee:.2f} 元")
print(f"折價券：{discount:.2f} 元")
print(f"應付總額：{total_amount:.2f} 元")
print(f"分攤人數：{people_number} 人")
print(f"每人費用：{per_person:.2f} 元")
print(f"每人預算：{budget_per_person:.2f} 元")
print(f"是否超過預算：{over_budget}")
print("====================================")
```
